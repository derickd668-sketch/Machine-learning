```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
```

```python
df.plot(kind="scatter",x="Number of Bedrooms",y="House Price'in thousands'")
```

```python
x=df.drop(columns="House Price'in thousands'")
```

```python
x.shape
```

```python
y=df[["House Price'in thousands'"]]
```

```python
y.shape
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)#splitting the data into train set and test set
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
model=LinearRegression()
```

```python
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
print("Mean Squared Error")
print(mean_squared_error(y_test,y_pred))
print("R squared")
print(r2_score(y_test,y_pred))
print("Mean Absolute Error")
print(mean_absolute_error(y_test,y_pred))
```
