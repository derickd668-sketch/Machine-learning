```python
class Circle:
    def __init__(self,radius):
        self.radius = radius
    def getPerimeter(self):
        return 2*3.142*self.radius
    def getArea(self):
        return 2*3.142*self.radius**2
c1=Circle(7)
print(c1.getPerimeter())
print(c1.getArea())
        
```

```python
c2=Circle(14)
print(c2.getPerimeter())
print(c2.getArea())
```

```python
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def display(self):
        print("Student Name:",self.name)
        print("Student Age:",self.age)
    def study(self):
        print(self.name,"is studying Python.")
student1 = Student("Derick",20)
student2 = Student("Grace",22)
print(student1.name)
print(student1.age)
print(student2.name)
print(student2.age)
student1.display()
student1.study()
student2.display()
student2.study()
    
```

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance
    def deposit(self,amount):
        if amount > 0:
            self.__balance += amount
    def get_balance(self):
        return self.__balance
acc = BankAccount(1000)
acc.deposit(500)
print("Balance:",acc.get_balance())
```

```python
class Student:
    def __init__(self,name,age,course):
       self.name=name
       self.age=age
       self.course=course  #instance variables
    def display_details(self):
        print("Student Name:",self.name)
        print("Student Age:",self.age)
        print("Student course:",self.course)   #instance method
student1=Student("Derick",20,"Data Science")
student2=Student("Grace",21,"Computer Science")   #Object creation
print("Accessing instance variables")
print(student1.name)
print(student1.age)
print(student2.course)
print()
print("Student 1 Details")
student1. display_details()
print()
print("Student 2 Details")
student2.display_details()

        
```
