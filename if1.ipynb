```python
sales=int(input("Enter the sales"))
if(sales<=5000):
    commission=0.05
elif(sales>5000 and sales<=10000):
    commission=0.1
elif(sales>10000):
    commission=0.15
else:
    print("Invalid sales")
total_commission=sales*commission
print(total_commission)
print("Thank you for working with us")
```
