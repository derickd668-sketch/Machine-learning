```python
#program to calculate the commision earned  by sales persons
sales=int(input("Enter The amount of sales"))
if(sales>=20000):
    commission = 0.2
elif(sales>=15000 and sales<20000):
    commission = 0.15
elif(sales>=10000 and sales<15000):
    commission = 0.1
elif(sales>=5000 and sales<10000):
    commission = 0.05
else:
    print("No commission")
total_commission=(sales*commission)
print(total_commission)

```
