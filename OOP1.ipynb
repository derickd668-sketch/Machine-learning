```python
class Employee:
    def __init__(self):
        self.__salary = 0
    def set_salary(self, salary):
        if salary >0:
            self.__salary = salary
        else:
            print("Salary must be greater than 0")
    def get_salary(self):
        return self.__salary
emp = Employee()
emp.set_salary(5000)
print("Salary:",emp.get_salary())
    
```

```python
emp.set_salary(-200)
```

```python
class BankAccount:
    def __init__(self,balance):
        self.__balance=balance
    def display(self):
        return self.__balance
    def add(self,amount):
        self.__balance += amount
obj1=BankAccount(5000)
print("The current balance is:",obj1.display())
obj1.add(700)
print("New balance is:",obj1.display())
        
    
```

```python
class Student:
    def __init__(self,name,marks):
      self.name=name
      self.__marks=marks
    def display(self):
        return self.__marks
obj1=Student("Derick",50)
print("My name is:",obj1.name,", and my marks is:",obj1.display())
```

```python
class Person:
    def __init__(self,name,idc,age):
        self.name=name
        self.idc=idc
        self.age=age
    def display_info(self):
        return f"The name is:{self.name},The ID is:{self.idc} and the age is{self.age}"
class Student(Person):
    def study(self):
        return f"Student, {self.name} ,is studying"
student1=Student("Mary",2345,24)
print(student1.display_info())
print(student1.study())
```

```python
class Animal:
    def __init__(self,name):
        self.name=name
    def speak(self):
        return f"{self.name}makes a sound"
class Dog(Animal):
    def speak(self):
        return f"{self.name} Woofs!"
class Cat(Animal):
    def speak(self):
        return f"{self.name} meaws!"
dog=Dog("Buddy")
dog.speak()
cat=Cat("Milky")
cat.speak()
```

```python
# Example: Inheritance
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(f"{self.name} makes a sound")

class Dog(Animal):
    def speak(self):
        print(f"{self.name} says Woof!")
class Cat(Animal):
    def speak(self):
        print(f"{self.name} says Meew!")
dog = Dog("Buddy")
dog.speak()
cat=Cat("Milky")
cat.speak()
```
