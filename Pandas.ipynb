```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))
data.boxplot(patch_artist=True,notch=True,boxprops=dict(facecolor='lightblue',color = 'blue'),
            medianprops=dict(color='red',linewidth=2),
             whiskerprops=dict(color='green'),
             flierprops=dict(marker='o',markerfacecolor='orange',markersize=6))
plt.title("Box Plot of Student Performance Dataset",fontsize=16)
plt.xlabel("Values",fontsize=12)
plt.xticks(rotation=15)
plt.grid(True, linestyle='--',alpha=0.7)
plt.show()
```
