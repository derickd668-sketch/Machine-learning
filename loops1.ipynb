```python
marks = int(input("Enter your marks(0-100):"))
if marks < 0 or marks > 100:
  print("Invalid marks entered.")
elif marks >= 80:
  print("Grade: A")
  print("Excellent!")
elif marks >= 70:
  print("Grade: B")
  print("Very Good!")
elif marks >= 60:
  print("Grade C")
  print("Good")
elif marks >= 50:
  print("Grade D")
  print("Pass")
else:
  print("Grade: E")
  print("Fail. Work harder.")

                    
```

```python
commission = 0
sales=int(input("Enter the sales:"))
if(sales<=5000):
    commission=0.05
elif(sales>5000 and sales <=10000):
    commission=0.1
elif(sales>10000):
    commission=0.15
else:
    print("Invalid Sales")
total_commission=sales*commission 
print(total_commission)
```
