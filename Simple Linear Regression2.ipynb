```python
import pandas as pd
data=pd.read_csv("Performance.csv")
print(data)
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import plotly.express as px
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv("Performance.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.plot(kind="scatter",x="Previous Scores",y="Current Performance")
```

```python
import plotly.express as px
Previous_Scores=[99,82,51,52,75,63,67,77,49,43]
Current_Performance=[98,89,50,47,77,61,69,73,50,37]
fig=px.bar(x=Previous_Scores,y=Current_Performance,title="Performance Analysis",labels={"x":"Previous Scores","y":"Current Performance"})
fig.show()
```

```python
x=df[["Previous Scores"]]   #I use the 2 opening and closing brackets because I want it to be in a data frame
y=df[["Current Performance"]]
```

```python
y.shape
```

```python
x.shape
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.4,random_state=42)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
model=LinearRegression()
```

```python
model.fit(y_train,x_train)
```

```python
y_pred=model.predict(y_test)
```

```python
plt.figure(figsize=(10,6))
plt.scatter(x=x_test,y=y_test,label="Actual Values",color="blue",marker="o")
plt.plot(x_test,y_pred,label="Predicted Values",color="red",marker="x")
```

```python
print(model.intercept_)
print(model.coef_)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
print("Mean Squared Error")
print(mean_squared_error(x_test,y_pred))
print("R_squared")
print(r2_score(x_test,y_pred))
print("Mean Absolute Error")
print(mean_absolute_error(x_test,y_pred))

```
